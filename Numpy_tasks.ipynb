Numpy Core Operations





```python
import numpy as np
```

```python
#Numpy List
values = np.array([34, 86, 90, 27, 66, 47])
print(values)
```

```python
#Dimensional Arrays
# one dimen... array
numbers = np.array([34, 86, 90, 27, 66, 47])
print(numbers)
# two dimen... array
numbers = np.array([
     [34, 86, 90],
     [44, 84, 70],
     [56, 62, 57]
])
print(numbers)
```

```python
#Important Array Properties
values = np.array([
    [34, 86, 90, 77],
    [44, 84, 70, 20],
    [56, 62, 57, 45]
])
print(values.ndim)
print(values.shape)
print(values.size)
print(values.dtype)
```

```python
#Indexing
numbers = np.array([34, 86, 90, 27, 66, 47])
print(numbers[5])
print(numbers[-3])
#indexing in a 2D array
values = np.array([
    [34, 86, 90, 77],
    [44, 84, 70, 20],
    [56, 62, 57, 45]
])
print(values[0,2])
```

```python
#Slicing
numbers = np.array([10, 20, 30, 40, 50])
print(numbers[1:3])
```

```python
#Vectorized Operations
numbers = np.array([34, 86, 90, 27, 66, 47])
print(numbers + 5)
print(numbers % 5)
print(numbers / 5)
print(numbers * 5)
print(numbers - 5)
```

```python
#Operation between IN 2 Arrays
numbers = np.array([10, 20, 30, 40, 50])
values = np.array([34, 86, 90, 66, 47])

print(numbers + values)
print(numbers / values)
print(numbers - values)
print(numbers * values)
print(numbers % values)
```

```python
#Mathematical Functions
numbers = np.array([14, 12, 22, 44])
print(np.sqrt(numbers))
print(np.power(numbers, 2))
numbers = np.array([-34, -86, -66, -42])
print(np.abs(numbers))
numbers = np.array([3.4, 8.6, 6.6, 4.2])
print(np.round(numbers))
```

```python
#Aggregation Functions
marks = np.array([14, 12, 22, 44])
print(np.sum(marks))
print(np.mean(marks))
print(np.max(marks))
print(np.min(marks))
print(np.median(marks))
print(np.std(marks))
```

```python
#Conditions in Numpy
marks = np.array([14, 12, 22, 44])
print(marks >= 60)
print(marks[marks>=40])
```

```python
#Filtering Data
marks = np.array([19, 80, 92, 45, 30, 20])
passed = marks[marks >= 40]
print(passed)
failed = marks[marks < 40]
print(failed)
```

```python
#Reshape
marks = np.array([19, 80, 92, 45, 30, 20])
new_array = marks.reshape(2, 3)
print(new_array)
```

```python
#Flatten
numbers = np.array([
     [34, 86, 90],
     [44, 84, 70],
     [56, 62, 57]
])
print(numbers.flatten())
```

```python
# Useful Numpy Functions
# 1.zeros()
print(np.zeros(5))
# 2.ones()
print(np.ones(12))
# 3.arrange()
print(np.arange(1, 14))
```

```python
# Random Numbers
marks = np.random.randint(30, 90, 20)
print(marks)
```
